# feature engineering


```python
df['boundary_runs'] = df['fours']*4 + df['sixes']*6
```

```python
df['strikeRate'] = pd.to_numeric(df['strikeRate'], errors='coerce')
df['sr_category'] = pd.cut(df['strikeRate'],
                          bins=[0,100,150,200,300],
                          labels=['Low','Medium','High','Very High'])
```

```python
if 'runs_per_ball' in df.columns:
    df['runs_per_ball'].fillna(0, inplace=True)
if 'boundary_pct' in df.columns:
    df['boundary_pct'].fillna(0, inplace=True)
```

```python
df['sr_category'] = df['sr_category'].cat.add_categories(['Unknown'])
df['sr_category'].fillna('Unknown', inplace=True)
```

```python
df.isnull().sum()
```

```python
df['sr_category'] = pd.cut(df['strikeRate'],
                          bins=[-1, 50, 100, 150, 200, 1000],
                          labels=['Very Low','Low','Medium','High','Very High'])
```

```python
plt.figure(figsize=(4,3))
sns.histplot(df['strikeRate'], kde=True)
plt.title("Strike Rate Distribution")
plt.show()

print("Insight: Most strike rates are in a moderate range.")
```

```python
plt.figure(figsize=(5,3))
sns.countplot(x='country', data=df)
plt.xticks(rotation=45)
plt.title("Matches by Country")
plt.show()

print("Insight: Shows where most matches are played.")
```

```python
plt.figure(figsize=(4,3))
sns.scatterplot(x='ballsFaced', y='runs', data=df)
plt.title("Runs vs Balls Faced")
plt.show()

print("Insight: More balls faced → more runs scored.")
```

```python
plt.figure(figsize=(4,3))
sns.scatterplot(x='strikeRate', y='runs', data=df)
plt.title("Strike Rate vs Runs")
plt.show()

print("Insight: High strike rate doesn’t always mean high runs.")
```

```python
plt.figure(figsize=(4,3))
sns.scatterplot(x='boundary_runs', y='runs', data=df)
plt.title("Boundary Runs vs Total Runs")
plt.show()

print("Insight: Boundaries contribute heavily to total runs.")
```

```python
plt.figure(figsize=(6,4))
sns.countplot(y='home_team', data=df,
              order=df['home_team'].value_counts().index)
plt.title("Matches by Team")
plt.show()

print("Insight: Shows most active teams.")
```

```python
top_players = df.groupby('fullName')['runs'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(6,4))
top_players.plot(kind='bar')
plt.title("Top 10 Players by Runs")
plt.xticks(rotation=45)
plt.show()

print("Insight: Identifies top-performing players.")
```

```python
plt.figure(figsize=(5,3))
df.groupby('season')['runs'].sum().plot(marker='o')
plt.title("Runs Over Seasons")
plt.show()

print("Insight: Shows performance trends over time.")
```

```python
plt.figure(figsize=(5,3))
sns.heatmap(df[['runs','ballsFaced','fours','sixes','strikeRate']].corr(),
            annot=True, cmap='coolwarm')

plt.title("Correlation Heatmap")
plt.show()

print("Insight: Runs strongly depend on balls faced and boundaries.")
```
